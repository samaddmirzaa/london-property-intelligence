```python
import pandas as pd
from pathlib import Path

root = Path(".").resolve().parent
pc = pd.read_csv(root / "data" / "supplementary" / "london_postcodes.csv")

# keep only the columns the app needs
pc = pc[["Postcode", "Latitude", "Longitude", "London zone",
         "Index of Multiple Deprivation", "Average Income", "Distance to station"]]
pc.columns = ["postcode", "lat", "lon", "zone", "imd_index", "avg_income", "dist_station"]
pc = pc.dropna(subset=["lat", "lon"])

# save as compact parquet
pc.to_parquet(root / "data" / "supplementary" / "postcodes_london.parquet", index=False)
print(f"Saved {len(pc):,} rows, {pc.shape[1]} columns")
```
